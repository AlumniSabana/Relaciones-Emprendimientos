```python
dane = pd.read_csv("https://www.datos.gov.co/resource/gdxc-w37w.csv?$limit=2000", dtype=str)

def bonito(s):
    return s.str.title().str.replace(r"(?<=\s)(De|Del|Y)\b", lambda m: m.group().lower(), regex=True)

dpto = dane.drop_duplicates("cod_dpto").set_index("cod_dpto")["dpto"]
dpto = pd.Series(bonito(dpto).values, index=dpto.index)   # '25' -> 'Cundinamarca'
mpio = dane.set_index("cod_mpio")["nom_mpio"]
mpio = pd.Series(bonito(mpio).values, index=mpio.index)   # '25175' -> 'Chía'
```

```python
df.columns = df.columns.str.strip()

basura = ["No informa", "NOINFORMA", "(Seleccione un valor)", "Despachos a todo el país"]
for col in ["Ciudad", "Departamento"]:
    df[col] = df[col].str.strip().replace(basura, pd.NA)

es_cod = lambda s: s.where(s.str.fullmatch(r"\d+", na=False))
cod_ciu = es_cod(df["Ciudad"]).str.zfill(5)        # '8001' -> '08001'
cod_dep = es_cod(df["Departamento"]).str.zfill(2)  # '5' -> '05'

df["Ciudad"] = cod_ciu.map(mpio).fillna(df["Ciudad"])
df["Departamento"] = cod_dep.map(dpto).fillna(df["Departamento"])
for col in ["Ciudad", "Departamento"]:
    df.loc[df[col].str.fullmatch(r"\d+", na=False), col] = pd.NA   # códigos que no existen en DANE (ej. '1')

# si falta el departamento, sale de los 2 primeros dígitos del código de ciudad
df["Departamento"] = df["Departamento"].fillna(cod_ciu.str[:2].map(dpto))

df["Departamento"] = df["Departamento"].replace({
    "San Andrés y Providencia": dpto["88"],
    "San Andrés, Providencia y Santa Catalina": dpto["88"],
})
df["Ciudad"] = df["Ciudad"].replace({
    "Bogotá": mpio["11001"],            # 'Bogotá, D.C.'
    "Cartagena": mpio["13001"],         # 'Cartagena de Indias'
    "San Andrés Isla": mpio["88001"],   # 'San Andrés'
})
```

```python
import csv
import pandas as pd

# Catálogo de actividades económicas (CIIU). Viene con cada fila entre comillas y con ';' partiendo
# algunas líneas, por eso se lee a mano en vez de con read_csv.
filas = []
with open("Catálogo_de_actividades_económicas_20261030.csv", encoding="utf-8-sig") as f:
    for linea in f:
        linea = linea.strip().rstrip(";").replace('";"', ";")
        filas.append(next(csv.reader([linea[1:-1].replace('""', '"')])))
ciiu = pd.DataFrame(filas[1:], columns=filas[0]).set_index("code")["description"].str.rstrip(".").str.replace("\u00ad", "")  # quita guiones blandos invisibles

# df_ok ya trae Servicio, Sector, Descripción y Razón Social por separado
base = df_ok.copy()
print("Catálogo:", ciiu.shape, "| Base:", base.shape)
```

```python
# Descripciones revisadas a mano (ortografía y redacción) de las empresas que lo necesitaban.
# Archivo: Descripciones_corregidas.csv. Descripción y Servicio se reemplazan por completo (vacío = sin dato);
# en Sector, "-" lo vacía (código de actividad mal declarado) y vacío lo deja igual.
corr = pd.read_csv("Descripciones_corregidas.csv", encoding="utf-8-sig", dtype=str, keep_default_na=False)
corr = corr.set_index("Eloqua Unique Identifier")
revisada = base["Eloqua Unique Identifier"].isin(corr.index)
for columna in ["Descripción", "Servicio que ofrece"]:
    nuevo = base["Eloqua Unique Identifier"].map(corr[columna]).replace("", pd.NA)
    base[columna] = nuevo.where(revisada, base[columna])
vaciar = base["Eloqua Unique Identifier"].map(corr["Sector"]).eq("-")
base.loc[vaciar, "Sector"] = pd.NA
print("Empresas con descripción revisada:", revisada.sum(), "de", len(base))
```

```python
basura = {"no informa", "noinforma", "(seleccione un valor)", "---", "-", "n/a", "na",
          "ninguno", "ninguna", "persona natural", "soy persona natural", "no aplica", "no tiene"}

def util(s):
    """Texto limpio; nulo si es relleno ('No informa', '---') o un código ('24', 'C1')."""
    s = s.str.strip().str.replace(r"\s+", " ", regex=True)
    malo = s.str.lower().isin(basura) | s.str.fullmatch(r"[CV]?\d+", na=False)
    return s.mask(malo)

fin = lambda s: s.str.rstrip(" .;,!")   # quita puntuación sobrante al final

desc = fin(util(base["Descripción"]))
serv = fin(util(base["Servicio que ofrece"]))
rs = fin(util(base["Razón Social"]))

# Sector: los códigos de 3-4 dígitos son CIIU y se traducen con el catálogo. Los de 1-2 dígitos son ids del
# formulario sin significado claro, y 1313 aparece en empresas de rubros no relacionados: no se traducen.
# La categoría (C1-C12) se eliminó al inicio del procesado: el CIIU de Sector la reemplaza.
sec = base["Sector"].str.strip()
cod = sec.where(sec.str.fullmatch(r"\d{3,4}", na=False) & (sec != "1313")).str.zfill(4)
actividad = cod.map(ciiu)
sector_txt = util(base["Sector"])   # los pocos sectores que ya vienen en texto ('Educación', 'ONG')

# no repetir lo que ya dice la descripción
rs = rs.mask(rs.str.lower() == base["Nombre de empresa"].str.strip().str.lower())
serv = serv.mask([isinstance(s, str) and isinstance(d, str) and s.lower() in d.lower()
                  for s, d in zip(serv, desc)])

partes = pd.concat([
    desc + ".",
    "Ofrece: " + serv + ".",
    "Actividad económica registrada: " + actividad + " (CIIU " + cod + ").",
    "Sector: " + sector_txt + ".",
    "Razón social: " + rs + ".",
], axis=1)

# todas las descripciones están revisadas a mano (ortografía y mayúsculas): solo se asegura la mayúscula
# inicial de cada parte de la frase, sin tocar el resto
inicial = lambda s: s.str[:1].str.upper() + s.str[1:]
texto = partes.apply(inicial).apply(lambda r: " ".join(r.dropna()), axis=1).str.replace("\u00ad", "")
base["Descripción completa"] = texto.replace("", pd.NA)

# Sector queda convertido: nombre de la actividad CIIU, o el texto original si ya era texto; los demás códigos quedan vacíos
base["Sector"] = actividad.fillna(sector_txt)
```

```python
# NIT dañado por Excel (notación científica, p. ej. 9.01013E+11): no se puede recuperar, se deja vacío
base.loc[base["ID de Empresa (NIT)"].str.contains(r"E\+", na=False), "ID de Empresa (NIT)"] = pd.NA

# solo las empresas con Descripción completa
base = base[base["Descripción completa"].notna()].copy()
print("Con Descripción completa:", len(base))

# Empresas repetidas: de cada grupo se conserva la fila con más campos con dato (a igualdad, la de texto más largo)
base = base.assign(_campos=base.notna().sum(axis=1), _largo=base["Descripción completa"].str.len())
base = base.sort_values(["_campos", "_largo"], ascending=False)

nombre = base["Nombre de empresa"].str.strip().str.lower().str.replace(r"\s+", " ", regex=True)
correo = base["Dirección de correo electrónico"].str.strip().str.lower()
nit9 = base["ID de Empresa (NIT)"].str.replace(r"\D", "", regex=True).str[:9]   # sin dígito de verificación

# 1) mismo nombre ('no' y 'no tiene nombre como tal' son relleno, no una empresa)
quitar = nombre.duplicated() & ~nombre.isin(["no", "no tiene nombre como tal"])

# 2) mismo NIT y (mismo correo, o un nombre contenido en el otro): la misma empresa escrita distinto
# shortcut: empresas con el mismo NIT, distinto nombre y distinto correo se dejan, revisar a mano si molestan
for _, g in base[(nit9.str.len() >= 8) & ~quitar].groupby(nit9):
    conservadas = []
    for i in g.index:
        if any(correo[i] == correo[j] or nombre[i] in nombre[j] or nombre[j] in nombre[i] for j in conservadas):
            quitar[i] = True
        else:
            conservadas.append(i)

print("Filas repetidas que se quitan:", quitar.sum())
print(base.loc[quitar, ["Nombre de empresa", "ID de Empresa (NIT)", "Dirección de correo electrónico"]].to_string())
base = base[~quitar].drop(columns=["_campos", "_largo"]).sort_index()
```

```python
# Sector relacionado: grupo de actividad (16 grupos) asignado con IA a partir de "Descripción completa".
# El resultado está en Sector_relacionado_ia.csv (una fila por empresa, unida por el ID de Eloqua).
rel = pd.read_csv("Sector_relacionado_ia.csv", encoding="utf-8-sig", dtype=str).set_index("Eloqua Unique Identifier")["Sector relacionado"]
base["Sector relacionado"] = base["Eloqua Unique Identifier"].map(rel)
print("Empresas sin Sector relacionado:", base["Sector relacionado"].isna().sum())
base["Sector relacionado"].value_counts()
```
